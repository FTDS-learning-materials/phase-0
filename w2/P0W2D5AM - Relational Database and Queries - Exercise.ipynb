| product_name | price | quantity | sales_class
| --- | --- | --- | ---
| Coffee Mug | 15.00 | 50 | Low
| Notebook | 5.00 | 100 | Low
| Pen | 2.50 | 200 | Low
| Gaming Chair | 150.00 | 1 | Low
| Headphones | 200.00 | 10 | Medium
| Smartphone | 800.00 | 8 | High

**Note**:
- `total_cost` is `price` x `quantity`
- If `total_cost` is higher than 5000 -> `sales_class` is **High**
- If `total_cost` is between 1000 to 5000 -> `sales_class` is **Medium**
- If `total_cost` is lower than 1000 -> `sales_class` is **Low**
