## B. Create a DataFrame and a Series

### B.1 - DataFrame

**A DataFrame is a 2-dimensional tabular data structure in Pandas.** You can create one from various sources, including dictionaries, lists, or external files like `CSV`.

```python
# Creating a DataFrame from a dictionary

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 22],
    'City': ['New York', 'London', 'Paris']}

df = pd.DataFrame(data)

print(df)
```

### B.2 - Series

**A Series is a one-dimensional labeled array in Pandas.** You can create one from a list, dictionary, or an ndarray.

```python
# Creating a Series from a list

data = [10, 20, 30, 40, 50]

series = pd.Series(data)

print(series)
```

### C.2 - From a `TSV` file  or other format file

`pd.read_csv` can also read any other extensions such as `.tsv`, `.txt`, but only if you specify the delimiter of the data.

The delimiter of `CSV` file is comma. Instead, the `TSV` file is tab or `\t`.

```python
# Open a TSV file using Pandas

tsv_df = pd.read_csv('https://raw.githubusercontent.com/FTDS-learning-materials/phase-0/refs/heads/v2.4/src/cities_world.tsv', delimiter='\t')
tsv_df
```

How about if we change the delimiter to default (',')?

```python
# Try open a TSV file using default delimiter

tsv_df2 = pd.read_csv('https://raw.githubusercontent.com/FTDS-learning-materials/phase-0/refs/heads/v2.4/src/cities_world.tsv')
tsv_df2
```

As you can see, the values ​​in each column cannot be separated.

### C.3 - From an Excel file

Excel file is special that has different encode, format to csv or other text files. Therefore, Pandas provides a particular function that is `pd.read_excel()`.

```python
# Open an Excel file using Pandas

excel_file = 'https://github.com/FTDS-learning-materials/phase-0/raw/refs/heads/v2.4/src/data_penjualan.xlsx'

sales_df = pd.read_excel(excel_file)
sales_df
```

Remember that excel file can possess more than one sheet. So, how to check sheets in the excel file?

```python
# List all sheets from an Excel file

pd.ExcelFile(excel_file).sheet_names
```

To open specific sheet, you can add `sheet_name` argument when use `pd.read_excel`. Note that by default, `pd.read_excel` will open the first sheet.

```python
# Open a specific sheet from an Excel file

sales_df_2021 = pd.read_excel(excel_file, sheet_name='2021')
sales_df_2021
```

## Query/Filtering

### Boolean Indexing

#### Single Condition

Retrieve data that has population more than 15 million people

```python
df[df['Population']>15000000]
```

#### Multiple Conditions

Cities that has more than 10 million of population in Southern Hemisphere

```python
df[(df['Population']>10000000) & (df['Latitude']<0)]
```

### Query Method

```python
df.query('Population > 15000000')
```

```python
df.query('Population > 10000000 and Latitude < 0')
```

### isin Method

Retrieve list City only in Argentina, Brazil, and Chile:

```python
df[df['Country'].isin(['Argentina','Brazil','Chile'])]
```

### str Method

Retrieve Cities that contains 'City' on their name:

```python
df[df['City'].str.contains('City')]
```

### .loc Method

```python
df.loc[(df['Population']>10000000)]
```

## Add Column and Row

### Add a New Column

```python
df['New_Pop'] = df['Land_area'] * df['Density']
df.head()
```

### Add a New Row

1. First Method : Using `.loc`

```python
# new row infomation
data = {
    'City': 'Surabaya',
    'Latitude': -7.2575,
    'Longitude': 112.7521,
    'Country': 'Indonesia',
    'Population': 2949585,
    'Land_area': 351,
    'Density': 8406,
    'Number':126
}

# use len() to put the new data at the last row
df.loc[len(df)] = data
df.tail()
```

2. Method 2: Using `concat()`

```python
# new row infomation
data = {
    'City': 'Bandung',
    'Latitude': -6.9147,
    'Longitude': 107.6098,
    'Country': 'Indonesia',
    'Population': 2498657,
    'Land_area': 167,
    'Density': 14907,
    'Number':210
}

# create a dataframe as a new row
new_row = pd.DataFrame([data])

# concatenate the new row with the original dataframe
df = pd.concat([df, new_row], ignore_index=True)
df.tail()
```

## F. Data Grouping

### F.1 - Grouping Based On A Single Column

```python
# One aggregation function
df.groupby('Country').mean(numeric_only=True)  # Automatically uses numeric columns
```

```python
# Filter numeric columns before aggregation
numeric_cols = df.select_dtypes(include='number').columns
df.groupby('Country')[numeric_cols].agg(['count', 'mean']).head()
```

```python
df.groupby('Country').agg({'City':'count','Population':'mean','Density':'max'}).head()
```
